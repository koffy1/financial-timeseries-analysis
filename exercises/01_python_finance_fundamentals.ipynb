```python
pos = {'AAPL':1000,'TSLA':500,'BAC':-250,'GS':1200}
ret = {'AAPL':-0.01,'TSLA':0.05,'BAC':0.01,'GS':0.03}

def compute_profit(pos,ret):
    profit={}
    for key in pos:
        profit[key] = pos[key] * ret[key]
    return profit

compute_profit(pos,ret)
```

```python
# Generate random returns. 
# Don't worry about how to generate these for now.
import numpy as np 
np.random.seed(2)
rets = list(np.random.normal(0, 0.1/np.sqrt(252), 252))
```

```python
def compute_stats(rets):
    stats = {}
    stats['avg'] = sum(rets)/len(rets)
    positives = 0
    for x in rets:
        if x > 0:
            positives += 1
    stats['hit_rate'] = positives / len(rets) * 100
    stats['max_ret'] = max(rets)
    return stats
compute_stats(rets)
```

```python
# Generate random returns and a signal.
# Don't worry about how to generate these for now.
import numpy as np 
np.random.seed(2)
rets = list(np.random.normal(0, 0.1/np.sqrt(252), 252))
signal = list(np.random.normal(0,1,252))
```

```python
def analyze_signal(rets,signal):
    analysis = {}
    pos_rets = 0
    pos_count = 0
    neg_rets = 0
    neg_count = 0

    for i in range(len(rets)):
        if signal[i] > 1:
            pos_rets += rets[i]
            pos_count += 1

        elif signal[i] < -1:
            neg_rets += rets[i]
            neg_count += 1

    analysis['pos_ret'] = pos_rets / pos_count
    analysis['neg_ret'] = neg_rets / neg_count
    analysis['spread'] = analysis['pos_ret'] - analysis['neg_ret']
    return analysis
analyze_signal(rets,signal)
```
